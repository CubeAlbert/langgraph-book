```python
from typing import TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name: str
    values: list[int]
    result: str
```

```python
def process_values(state: AgentState) -> AgentState:
    """This function handles multiple different inputs"""
    state['result'] = f"Hi, {state['name']}, the sum is {sum(state['values'])}"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("process", process_values)
graph.set_entry_point("process")
graph.set_finish_point("process")

app = graph.compile()

result = app.invoke({"name":"Bob", "values": [1,2,3,4]})
print(result)
```

```python
class MyState(TypedDict):
    name: str
    values: list[int]
    operation: str
    result: str


def mul(input: list[int]) -> int:
    result = 1
    for x in input:
        result *= x
    return result


def calc(input: MyState) -> MyState:
    if input["operation"] == "*":
        calc_result = mul(input["values"])
    else:
        calc_result = sum(input["values"])

    input["result"] = f"Hi {input["name"]}, your answer is: {calc_result}"
    return input


graph = StateGraph(MyState)

graph.add_node("process", calc)
graph.set_entry_point("process")
graph.set_finish_point("process")

my_app = graph.compile()

result = my_app.invoke(
    {"name": "Jack Sparrow", "values": [1, 2, 3, 4], "operation": "*"}
)

print(result["result"])
```
