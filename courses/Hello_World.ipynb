# Hello World Agent

Our first LangGraph graph!

```python
from typing import Dict, TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    message: str

def greeting_node(state: AgentState) -> AgentState:
    """Simple node that returns a greeting message."""
    state["message"] = "Hello! "+ state["message"] + ", How can I assist you today?"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("greeting", greeting_node)
graph.set_entry_point("greeting")
graph.set_finish_point("greeting")

app = graph.compile()
```

```python
input = AgentState(message="Bob")
output = app.invoke(input)
print(input["message"])   
print(output["message"])  
```

```python
class MyState(TypedDict):
    message: str


def gen_output(input: MyState) -> MyState:
    output: MyState = {"message": input["message"] + ", you are doing well."}
    return output


my_graph = StateGraph(MyState)

my_graph.add_node("gen_output", gen_output)
my_graph.set_entry_point("gen_output")
my_graph.set_finish_point("gen_output")

my_app = my_graph.compile()

bob = MyState(message="Bob")
bob_result: MyState = my_app.invoke(bob)
print(bob_result["message"])
```
